```python
import pandas as pd

data = {
    "Sky": ["Sunny", "Sunny", "Rainy", "Sunny"],
    "Temperature": ["Warm", "Warm", "Cold", "Warm"],
    "Humidity": ["Normal", "High", "High", "High"],
    "Wind": ["Strong", "Strong", "Strong", "Strong"],
    "EnjoySport": ["Yes", "Yes", "No", "Yes"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Start with the first positive example
hypothesis = None

for index, row in df.iterrows():

    if row["EnjoySport"] == "Yes":

        if hypothesis is None:
            hypothesis = list(row[:-1])

        else:
            for i in range(len(hypothesis)):
                if hypothesis[i] != row.iloc[i]:
                    hypothesis[i] = "?"

print("Final Hypothesis:", hypothesis)
```

```python
attributes = list(df.columns[:-1])

print("Final Most Specific Hypothesis:")

for attribute, value in zip(attributes, hypothesis):
    print(attribute, "=", value)
```
